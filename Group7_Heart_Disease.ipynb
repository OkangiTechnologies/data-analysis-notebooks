CREATING A DATA FRAME FROM A DICTIONARY

```python
# Create a dictionary containing student information
students_dictionary = {
    "ID": [1, 2, 3, 4, 5],
    "Name": ["John", "Mary", "Peter", "Asha", "David"],
    "Age": [20, 21, 19, 22, 20]
}

# Convert the dictionary into a DataFrame
students_dict_df = pd.DataFrame(students_dictionary)

# Display the DataFrame
students_dict_df
```

display the columns:

```python
students_dict_df.columns
```

Display the data types:

```python
students_dict_df.dtypes
```

DataFrame from Series

```python
# Create three Pandas Series
student_names = pd.Series(["John", "Mary", "Peter", "Asha", "David"])
student_ages = pd.Series([20, 21, 19, 22, 20])
student_scores = pd.Series([75, 82, 68, 91, 79])

# Combine the Series into a DataFrame
students_series_df = pd.DataFrame({
    "Name": student_names,
    "Age": student_ages,
    "Score": student_scores
})

# Display the DataFrame
print(students_series_df)
```
