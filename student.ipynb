```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

students = {
    "Name": ["John", "Asha", "Peter", "Mary", "James",
             "Grace", "David", "Anna", "Michael", "Sarah"],

    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],

    "Course": ["Data Science", "Data Science", "Computer Science",
               "Data Science", "Computer Science", "Data Science",
               "Computer Science", "Data Science", "Computer Science",
               "Data Science"],

    "Attendance": [85, 92, 70, 95, 80, 88, 75, 90, 68, 94],

    "Assignment": [78, 88, 65, 90, 72, 84, 70, 91, 60, 87],

    "Exam": [82, 90, 60, 94, 75, 86, 68, 93, 58, 89]
}

df = pd.DataFrame(students)

print(df)
```

```python
print(df.head())
```

```python
print(df.describe())
print(df.info())
```

```python
average_exam=df["Exam"].mean()
print("The average Exam Score is:",average_exam)
highest_exam=df["Exam"].max()
print('The Highest score is :',highest_exam)


lowest_exam = df["Exam"].min()

print("Lowest Exam Score:", lowest_exam)
```

identify the best performing student

```python
best_student=df.loc[df["Exam"].idxmax()]
print("THE BEST PERFORMING STUDENT")
print(best_student)
```

CALCULATE AVERAGE ASSIGNMENT SCORE

```python
average_assignment = df["Assignment"].mean()

print("Average Assignment Score:", average_assignment)
```

CALCULATE AVERAGE ATTENDANCE

```python
average_attendance = df["Attendance"].mean()

print("Average Attendance:", average_attendance)
```

1️⃣ Create a Pass/Fail column

```python
df["Result"]=df["Exam"].apply(lambda score:"PASS" if score>=50 else "FAIL")
print(df)
```

 Count how many students passed

```python
count_passed=(df["Result"]=="PASS").sum()
print("THE NUMBER OF STUDENT PASSED",count_passed)
```

 Count how many students failed

```python
count_failed=(df["Result"]=="FAIL").sum()
print("THE NUMBER OF STUDENT WHO HAS FAILED IS:",count_failed)
```

Calculate the pass rate 

```python
pass_rate=(count_passed/len(df) )* 100
print("THE PASS RATE IS",pass_rate,"%")
```

Display only students who passed

```python
failed_students = df[df["Result"] == "FAIL"]

print(failed_students)
```

calculate an overall score using:

Attendance = 20%
Assignment = 30%
Exam = 50%

```python
df["Overall_Score"] = (
    df["Attendance"] * 0.20 +
    df["Assignment"] * 0.30 +
    df["Exam"] * 0.50
)

print(df)
```

Find the top 3 students

```python
top_students = df.sort_values(
    by="Overall_Score",
    ascending=False
).head(3)

print("TOP 3 STUDENTS")
print(top_students[["Name", "Course", "Overall_Score"]])
```

 BI KPI Summary

```python
total_students = len(df)
average_score = df["Overall_Score"].mean()
pass_count = (df["Result"] == "PASS").sum()
fail_count = (df["Result"] == "FAIL").sum()
pass_rate = (pass_count / total_students) * 100

print("===== STUDENT PERFORMANCE BI =====")
print("Total Students:", total_students)
print("Average Overall Score:", round(average_score, 2))
print("Students Passed:", pass_count)
print("Students Failed:", fail_count)
print("Pass Rate:", round(pass_rate, 2), "%")
```

Let's compare students' overall performance.

```python
plt.figure(figsize=(10, 5))

plt.bar(
    df["Name"],
    df["Overall_Score"]
)

plt.title("Overall Student Performance")
plt.xlabel("Student")
plt.ylabel("Overall Score")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

Which course has the highest average performance?

```python
course_performance = df.groupby("Course")["Overall_Score"].mean()

print(course_performance)
```

```python
plt.figure(figsize=(8, 5))

course_performance.plot(kind="bar")

plt.title("Average Performance by Course")
plt.xlabel("Course")
plt.ylabel("Average Overall Score")

plt.xticks(rotation=0)

plt.tight_layout()

plt.show()
```

 Attendance vs Exam Performance 

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Attendance"],
    df["Exam"]
)

plt.title("Attendance vs Exam Performance")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Score")

plt.tight_layout()

plt.show()
```

6. Pass vs Fail 

```python
result_counts = df["Result"].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    result_counts,
    labels=result_counts.index,
    autopct="%1.1f%%"
)

plt.title("Student Pass/Fail Distribution")

plt.show()
```

Question 1: Which course performs better?

```python
course_performance = df.groupby("Course")["Overall_Score"].mean()

print("Average Performance by Course:")
print(course_performance)
best_course = course_performance.idxmax()
best_course_score = course_performance.max()

print("Best Performing Course:", best_course)
print("Average Score:", round(best_course_score, 2))
```

 Find students who need attention 

```python
students_at_risk = df[df["Overall_Score"] < 70]

print("Students Requiring Academic Attention:")
print(
    students_at_risk[
        ["Name", "Course", "Attendance", "Exam", "Overall_Score"]
    ]
)
```

 Identify students with low attendance 🚨

```python
low_attendance = df[df["Attendance"] < 75]

print("Students with Low Attendance:")
print(
    low_attendance[
        ["Name", "Course", "Attendance", "Exam"]
    ]
)
```

Find the relationship between attendance and exam score 

```python
correlation = df["Attendance"].corr(df["Exam"])

print("Attendance-Exam Correlation:", round(correlation, 2))
```

Create a Performance Category 

```python
def performance_category(score):
    if score >= 80:
        return "Excellent"
    elif score >= 70:
        return "Good"
    elif score >= 50:
        return "Average"
    else:
        return "Poor"


df["Performance_Category"] = df["Overall_Score"].apply(
    performance_category
)

print(
    df[
        ["Name", "Overall_Score", "Performance_Category"]
    ]
)
```

6. Count students in each category

```python
category_counts = df["Performance_Category"].value_counts()

print(category_counts)
plt.figure(figsize=(8, 5))

category_counts.plot(kind="bar")

plt.title("Student Performance Categories")
plt.xlabel("Performance Category")
plt.ylabel("Number of Students")

plt.xticks(rotation=0)

plt.tight_layout()

plt.show()
```

 Business Recommendations

Based on our findings, recommendations could include:

1. Improve attendance monitoring

If attendance is strongly associated with exam performance, institutions should monitor students with low attendance.

2. Support students at risk

Students with low overall scores can be identified early and provided with additional academic support.

3. Monitor course performance

The course with the lowest average score can be investigated to identify possible challenges.

4. Recognize high performers

Students in the Excellent category can be recognized or rewarded.

5. Use BI dashboards for continuous monitoring

Instead of waiting until the end of a semester, administrators can monitor performance regularly.
